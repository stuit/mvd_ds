```python
# Function to group similar names
def group_similar(series, threshold=80):
	inds = pd.Series(series.index)
	norm_inds = inds.str.lower()
	# Normalize names
	s_normalized = norm_inds.apply(unidecode)
		
	groups = defaultdict(list)
	used = set()

	for i, name in enumerate(norm_inds):
		if name in used:
			continue
		groups[name].append(name)
		for j, other in enumerate(norm_inds):
			if i != j and fuzz.ratio(name, other) >= threshold:
				groups[name].append(other)
				used.add(other)

	return {k: list(set(v)) for k, v in groups.items()}

# Group similar names
# similar_names = group_similar(jqb_cnts, threshold=80)
similar_names = group_similar(mg_cnts, threshold=80)

max_length = max(len(v) for v in similar_names.values())
data = {k: v + [None] * (max_length - len(v)) for k, v in similar_names.items()}
jqb_posi_dict_df = pd.DataFrame(data)
print(jqb_posi_dict_df)
```

```python
# Function to group similar names
def group_similar(series, threshold=80):
    groups = defaultdict(list)
    used = set()
    
    for i, name in enumerate(series):
        if name in used:
            continue
        groups[name].append(name)
        for j, other in enumerate(series):
            if i != j and fuzz.ratio(name, other) >= threshold:
                groups[name].append(other)
                used.add(other)

    return {k: list(set(v)) for k, v in groups.items()}

inds = pd.Series(col30_cnts.index)
norm_inds = inds.str.lower()
# Normalize names
s_normalized = norm_inds.apply(unidecode)

# Group similar names
similar_names = group_similar(norm_inds, threshold=80)

print(similar_names)
```

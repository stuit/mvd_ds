```python
import pandas as pd
```

```python
def decode_text(text):
    """Decode text from 'latin1' to 'utf-8'."""
    return text.encode('latin1').decode('utf-8')

def load_and_decode_csv(file_path):
    """Load and decode a CSV file, then parse it into a pandas DataFrame."""
    # Load the CSV file with 'latin1' encoding
    df = pd.read_csv(file_path, encoding='latin1')
    
    # Decode each string in the DataFrame
    for column in df.columns:
        df[column] = df[column].apply(lambda x: decode_text(x) if isinstance(x, str) else x)
    
    return df

# Example usage
file_path = 'select.csv'
df = load_and_decode_csv(file_path)
print(df.head())
```

```python
date_reformated = df.copy()
date_reformated['R5'] = pd.to_datetime(df['R5']).dt.strftime('%d.%m.%Y')
date_reformated.head()
```

```python
df_full = pd.read_excel('recidivist.xlsx')
df_full.head()
```

```python
def find_unique_rows(df1, df2):
    # Merge df1 with df2 using a left join to keep all rows from df1
    merged_df = df1.merge(df2, on=['ФАМИЛИЯ', 'ИМЯ', 'ОТЧЕСТВО', 'ДАТА_РОЖДЕНИЯ'], how='left', indicator=True)
    # Filter rows where df2 didn't have a matching row (i.e., '_merge' column is 'left_only')
    unique_df = merged_df[merged_df['_merge'] == 'left_only'].drop(columns=['_merge'])
    return unique_df

date_reformated.rename(columns={'R2':'ФАМИЛИЯ','R3':'ИМЯ','R4':'ОТЧЕСТВО','R5':'ДАТА_РОЖДЕНИЯ'}, inplace=True)
# Find unique rows in df1 that are not in df2
unique_df = find_unique_rows(df_full, date_reformated)
print(unique_df)
```
